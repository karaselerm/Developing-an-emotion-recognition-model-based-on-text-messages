```python
import nest_asyncio
nest_asyncio.apply()
import logging
from telegram import Update
from telegram.ext import Application, CommandHandler, MessageHandler, filters, CallbackContext
from transformers import pipeline
import asyncio

# Настройка логирования для отслеживания ошибок
logging.basicConfig(format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
                    level=logging.INFO)
logger = logging.getLogger(__name__)

# Загружаем модель для анализа эмоций
emotion_classifier = pipeline("text-classification", model=r"C://Users//a.karakuchukova//Desktop//Университет//Курсовая работа//models//roberta-base_emotion_model")

# Маппинг эмоций
emotion_mapping_inverse = {
    'LABEL_0': 'neutral',
    'LABEL_1': 'happiness',
    'LABEL_2': 'sadness',
    'LABEL_3': 'anger',
    'LABEL_4': 'disgust',
    'LABEL_5': 'surprise',
    'LABEL_6': 'excited',
    'LABEL_7': 'fear'
}

# Функция обработки команды /start
async def start(update: Update, context: CallbackContext):
    await update.message.reply_text("Привет! Напиши мне сообщение на английском, и я скажу, какие эмоции в нем содержатся.")

# Функция обработки текстовых сообщений
async def classify_emotion(update: Update, context: CallbackContext):
    # Получаем текст сообщения
    user_message = update.message.text

    # Классификация эмоции с использованием модели
    emotions = emotion_classifier(user_message)

    # Если модель вернула результат
    if emotions:
        label = emotions[0]['label']  # Получаем метку, например 'LABEL_1'
        emotion = emotion_mapping_inverse.get(label, 'Неизвестная эмоция')  # Преобразуем метку в название эмоции
        confidence = emotions[0]['score']  # Уверенность в предсказании
        await update.message.reply_text(f"Эмоция в вашем сообщении: {emotion} с уверенностью {confidence:.2f}")
    else:
        await update.message.reply_text("Не могу определить эмоцию. Попробуйте переформулировать сообщение.")

# Основная функция
def main():
    API_TOKEN = "8035403699:AAFL5sM_y4Mi0QLQgaOAWpCj658MEODVNpw"  # Замените на ваш токен

    application = Application.builder().token(API_TOKEN).build()

    application.add_handler(CommandHandler("start", start))

    application.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, classify_emotion))

    asyncio.run(application.run_polling())

if __name__ == '__main__':
    main()
```
